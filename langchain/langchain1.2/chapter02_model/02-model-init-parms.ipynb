# 模型初始化的参数使用情况
## 1、temperature的使用
控制输出随机性，范围 0.0-2.0，温度越高输出越随机。
- 0.0 ：最确定性，输出几乎不变
- 1.0 ：平衡创造性和一致性
- 2.0 ：最随机，最有创造性

举例1 创作诗歌 temperature设置1.9

```python
import os
from pydantic import SecretStr
from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

# 1、读取.env配置文件中的信息。相关的环境变量以.env文件优先
load_dotenv(verbose=True)
OPENROUTER_API_KEY=SecretStr(os.getenv("OPENROUTER_API_KEY")or"")

# 2、模型的初始化
model=ChatOpenRouter(
    model="poolside/laguna-s-2.1:free",
    temperature=1.9,
    api_key=OPENROUTER_API_KEY,

)
# 3、模型的调用
for i in range(3):
    print(model.invoke("帮我写一首关于春天的七言绝句，用中文").content)
```

举例3 创意文案与头脑风暴 设置为1.5

```python
import os
from pydantic import SecretStr
from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

# 1、读取.env配置文件中的信息。相关的环境变量以.env文件优先
load_dotenv(verbose=True)
OPENROUTER_API_KEY=SecretStr(os.getenv("OPENROUTER_API_KEY")or"")

# 2、模型的初始化
model=ChatOpenRouter(
    model="poolside/laguna-s-2.1:free",
    temperature=1.5,
    api_key=OPENROUTER_API_KEY,

)
# 3、模型的调用

# 向模型发送单条数据
response = model.invoke("请为一款极致静音的机械键盘写3个充满诗意且极具张力的广告，用中文")
print(response.content)
```

## 2、max_tokens参数的使用
限制模型输出的最大 token 数量

```python
import os
from pydantic import SecretStr
from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

# 1、读取.env配置文件中的信息。相关的环境变量以.env文件优先
load_dotenv(verbose=True)
OPENROUTER_API_KEY=SecretStr(os.getenv("OPENROUTER_API_KEY")or"")

# 2、模型的初始化
model=ChatOpenRouter(
    model="poolside/laguna-s-2.1:free",
    max_tokens=15,
    api_key=OPENROUTER_API_KEY,

)
# 3、模型的调用

# 向模型发送单条数据
response = model.invoke("介绍一下你自己")
print(response.content)
```
